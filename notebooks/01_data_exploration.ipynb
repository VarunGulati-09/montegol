```python
df["date"] = pd.to_datetime(df["date"])

modern_df = df[df["date"] >= "2010-01-01"]

print(modern_df.shape)
```

```python
def get_result(row):
    if row["home_score"] > row["away_score"]:
        return "H"
    elif row["home_score"] < row["away_score"]:
        return "A"
    return "D"

modern_df["result"] = modern_df.apply(get_result, axis=1)

modern_df["result"].value_counts()
```

```python
future_matches = df[
    (df["home_score"].isna()) |
    (df["away_score"].isna())
]

print(future_matches.shape)

future_matches.head(20)
```

```python
future_matches["date"].min(), future_matches["date"].max()
```

```python
completed_matches = df[
    df["home_score"].notna() &
    df["away_score"].notna()
].copy()


future_fixtures = df[
    df["home_score"].isna() | 
    df["away_score"].isna()
].copy()
```

```python
print(completed_matches.shape)
print(future_fixtures.shape)
```

```python
future_teams = set(future_fixtures["home_team"]).union(
    set(future_fixtures["away_team"])
)

print("Number of teams:", len(future_teams))

for team in sorted(future_teams):
    print(team)
```

```python
future_fixtures["date"].value_counts().sort_index()
```

```python
for team in sorted(future_teams):
    count = len(
        completed_matches[
            (completed_matches["home_team"] == team) |
            (completed_matches["away_team"] == team)
        ]
    )
    print(team, count)
```
